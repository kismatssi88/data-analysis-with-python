Hypothesis1:Technology products have the highest profit margin compared to other product categories

```python
total_profit =df.groupby('Category')['Profit'].sum()
print(total_profit)
#visualization of total profit by category
total_profit.plot(kind='bar')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.title('Total Profit by Category')
plt.savefig('../images/super market/total_profit_by_category.png', dpi=300, bbox_inches='tight')
plt.show()
#conclusion:yes the Technology category is the most profitable category in the dataset.
```

Hypothesis2:The East region has the highest sales compared to other regions

```python
sales_by_region = df.groupby('Region')['Sales'].sum().round(2)
print(sales_by_region)
sales_by_region.plot(kind='bar')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.title('Total Sales by Region')
plt.savefig('../images/super market/sales_by_region.png', dpi=300, bbox_inches='tight')
plt.show()
#conclusion:no the central region has the highest sales in the dataset.
```

Hypothesis3:Sales are higher during certain months of the year

```python
df['Order Month']=pd.DatetimeIndex(df['Order Date']).month

month_sales=df.groupby('Order Month')['Sales'].sum()
month_sales.plot(kind='line')
plt.title("Total Sales by Month")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.savefig('../images/super market/total_sales_by_month.png', dpi=300, bbox_inches='tight')
plt.show()
#conclusion:yes the sales are higher in certain months of the year.
```

14.univariatve analysis

how many peoples survived and how many died?


```python
survived = (df["Survived"] == 1).sum()
died = (df["Survived"] == 0).sum()

print("Survived:", survived)
print("Died:", died)
```

```python
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt

counts = df["Survived"].value_counts()

plt.bar(["Died", "Survived"], counts)
plt.xlabel("Status")
plt.ylabel("Number of Passengers")
plt.title("Titanic Survival Count")
plt.savefig("../images/titanic/survival_count.png", dpi=300, bbox_inches="tight")
plt.show()
```

how many were in first class ,second class and third class

```python
first_class=(df['Pclass']==1).sum()
print("first_class:",first_class)
second_class=(df['Pclass']==2).sum()
print('second_class:',second_class)
third_class=(df['Pclass']==3).sum()
print('third class:',third_class)
```

```python
import matplotlib.pyplot as plt
counts = df["Pclass"].value_counts()

plt.bar(["first_class", "second_class","third_class"], counts)
plt.xlabel("class")
plt.ylabel("Number of Passengers")
plt.title("Titanic class Count")
plt.savefig("../images/titanic/class_count.png", dpi=300, bbox_inches="tight")
```
